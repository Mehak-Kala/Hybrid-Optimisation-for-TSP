```python
import tsplib95
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.distributions import Categorical
import matplotlib.pyplot as plt
import pandas as pd
import random
import time
from copy import deepcopy
import os

# ==========================================
# 0. CONFIGURATION & EMBEDDED DATA
# ==========================================
# Fallback data (Berlin52) so code runs immediately even if file is missing
BERLIN52_EMBEDDED = """NAME: berlin52
TYPE: TSP
COMMENT: 52 locations in Berlin (Groetschel)
DIMENSION: 52
EDGE_WEIGHT_TYPE: EUC_2D
NODE_COORD_SECTION
1 565.0 575.0
2 25.0 185.0
3 345.0 750.0
4 945.0 685.0
5 845.0 655.0
6 880.0 660.0
7 25.0 230.0
8 525.0 1000.0
9 580.0 1175.0
10 650.0 1130.0
11 1605.0 620.0
12 1220.0 580.0
13 1465.0 200.0
14 1530.0 5.0
15 845.0 680.0
16 725.0 370.0
17 145.0 665.0
18 415.0 635.0
19 510.0 875.0
20 560.0 365.0
21 300.0 465.0
22 520.0 585.0
23 480.0 415.0
24 835.0 625.0
25 975.0 580.0
26 1215.0 245.0
27 1320.0 315.0
28 1250.0 400.0
29 660.0 180.0
30 410.0 250.0
31 420.0 555.0
32 575.0 665.0
33 1150.0 1160.0
34 700.0 580.0
35 685.0 595.0
36 685.0 610.0
37 770.0 610.0
38 795.0 645.0
39 720.0 635.0
40 760.0 650.0
41 475.0 960.0
42 95.0 260.0
43 875.0 920.0
44 700.0 500.0
45 555.0 815.0
46 830.0 485.0
47 1170.0 65.0
48 830.0 610.0
49 605.0 625.0
50 595.0 360.0
51 1340.0 725.0
52 1740.0 245.0
EOF
"""

# Define content for eil51.tsp (a very small example, just for demonstration)
eil51_content = """NAME : eil51
TYPE : TSP
COMMENT : 51 cities (Eilon)
DIMENSION : 51
EDGE_WEIGHT_TYPE : EUC_2D
NODE_COORD_SECTION
1 37 52
2 49 49
3 52 64
4 20 26
5 40 30
6 21 47
7 17 63
8 31 62
9 52 33
10 51 21
11 42 41
12 31 32
13 5 25
14 12 42
15 36 16
16 52 41
17 27 23
18 17 33
19 13 13
20 57 58
21 52 62
22 47 62
23 20 62
24 38 41
25 39 49
26 43 45
27 25 39
28 29 33
29 45 42
30 20 48
31 52 24
32 50 16
33 39 27
34 32 37
35 34 50
36 34 39
37 20 40
38 52 30
39 31 16
40 50 20
41 53 26
42 39 10
43 31 29
44 50 56
45 45 57
46 51 58
47 28 35
48 40 69
49 47 50
50 50 49
51 51 31
EOF
"""

# Define content for st70.tsp (a very small example, just for demonstration)
st70_content = """NAME : st70
TYPE : TSP
COMMENT : 70 cities (Setubal)
DIMENSION : 70
EDGE_WEIGHT_TYPE : EUC_2D
NODE_COORD_SECTION
1 580 430
2 290 190
3 510 500
4 950 670
5 820 640
6 860 650
7 20 220
8 500 1000
9 570 1150
10 640 1100
11 1580 610
12 1190 560
13 1440 200
14 1500 0
15 820 670
16 700 350
17 120 650
18 390 620
19 490 850
20 540 340
21 280 440
22 500 560
23 460 390
24 810 600
25 940 560
26 1180 230
27 1290 300
28 1220 380
29 630 160
30 380 230
31 390 530
32 550 640
33 1120 1140
34 670 560
35 660 570
36 660 590
37 740 590
38 770 620
39 690 610
40 730 630
41 450 940
42 70 240
43 840 900
44 670 480
45 530 790
46 800 460
47 1140 40
48 800 590
49 580 600
50 570 340
51 1310 700
52 1710 230
53 430 870
54 480 940
55 490 910
56 500 880
57 510 850
58 520 820
59 530 790
60 540 760
61 550 730
62 560 700
63 570 670
64 580 640
65 590 610
66 600 580
67 610 550
68 620 520
69 630 490
70 640 460
EOF
"""

# Create the files in the /content/ directory if they don't exist
file_dir = '/content/'

eil51_path = os.path.join(file_dir, 'eil51.tsp')
if not os.path.exists(eil51_path):
    with open(eil51_path, 'w') as f:
        f.write(eil51_content)
    print(f"Created {eil51_path}")
else:
    print(f"{eil51_path} already exists.")

st70_path = os.path.join(file_dir, 'st70.tsp')
if not os.path.exists(st70_path):
    with open(st70_path, 'w') as f:
        f.write(st70_content)
    print(f"Created {st70_path}")
else:
    print(f"{st70_path} already exists.")


DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
print(f"Running Hybrid Optimization on: {DEVICE}")

class TSPManager:
    def __init__(self, filepath=None):
        # Tries to load file, falls back to embedded string if missing
        if filepath and os.path.exists(filepath):
            print(f"Loading data from file: {filepath}")
            self.problem = tsplib95.load(filepath)
        else:
            print("Target file not found. Loading embedded Berlin52 data...")
            self.problem = tsplib95.parse(BERLIN52_EMBEDDED)

        self.num_cities = self.problem.dimension
        nodes = list(self.problem.get_nodes())
        self.coords = np.array([self.problem.node_coords[n] for n in nodes])

        # Precompute Distance Matrix
        self.matrix = np.zeros((self.num_cities, self.num_cities))
        for i in range(self.num_cities):
            for j in range(self.num_cities):
                if i != j:
                    self.matrix[i][j] = np.linalg.norm(self.coords[i] - self.coords[j])

        # Dictionary of known optimal values for common TSPLIB instances
        KNOWN_OPTIMAL_VALUES = {
            'berlin52': 7542.0,
            'eil51': 426.0,
            'pr107': 44303.0,
            'kroA100': 21282.0,
            'eil76': 538.0,
            'st70': 675.0,
        }

        # Attempt to get optimal value
        # First, try to get it from a known dictionary based on problem name
        if self.problem.name in KNOWN_OPTIMAL_VALUES:
            self.optimal_value = KNOWN_OPTIMAL_VALUES[self.problem.name]
        # Fallback if not found in dictionary
        else:
            print(f"Warning: Optimal value not found for {self.problem.name}. Using placeholder 0.0 for gap calculation.")
            self.optimal_value = 0.0 # Placeholder if no optimal value is found

    def evaluate(self, route):
        """Calculates total tour distance (closed loop)"""
        d = 0.0
        for i in range(len(route)-1):
            d += self.matrix[route[i], route[i+1]]
        d += self.matrix[route[-1], route[0]]
        return d

# ==========================================
# 1. RL MODULE: PPO (Proximal Policy Optimization)
# ==========================================
class PPOActorCritic(nn.Module):
    def __init__(self, num_cities, hidden_dim=128):
        super(PPOActorCritic, self).__init__()
        # Input: One-Hot City + Visited Mask
        input_dim = num_cities * 2

        self.shared = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU()
        )

        self.actor = nn.Linear(hidden_dim, num_cities)
        self.critic = nn.Linear(hidden_dim, 1)

    def forward(self, x, mask):
        feat = self.shared(x)
        val = self.critic(feat)
        logits = self.actor(feat)

        # Mask visited nodes (log prob -> -inf)
        logits = logits.masked_fill(mask.bool(), -1e9)
        probs = F.softmax(logits, dim=-1)
        return probs, val

class PPOAgent:
    def __init__(self, env, lr=1e-3):
        self.env = env
        self.model = PPOActorCritic(env.num_cities).to(DEVICE)
        self.optimizer = optim.AdamW(self.model.parameters(), lr=lr)
        self.gamma = 0.99
        self.eps_clip = 0.2
        self.k_epochs = 4

    def generate_population(self, population_size=100, episodes=400):
        """
        Runs the RL training loop and returns the best 'population_size' routes found.
        These serve as the high-quality initial population for the GA.
        """
        print(f"  [RL Module] Training PPO for {episodes} episodes to generate population...")
        candidates = []
        best_rl_score = float('inf')

        for ep in range(episodes):
            tour, dist, memory = self._run_episode()

            # Batch update every 20 episodes
            if ep % 20 == 0 and ep > 0:
                self._update_policy([memory])

            # Save candidate
            candidates.append(tour)
            if dist < best_rl_score:
                best_rl_score = dist

            if ep % 100 == 0:
                print(f"    RL Episode {ep}: Best Score {best_rl_score:.2f}")

        # Return the top unique routes found by RL (sorted by distance)
        candidates.sort(key=lambda x: self.env.evaluate(x))
        return candidates[:population_size], best_rl_score

    def _run_episode(self):
        n = self.env.num_cities
        curr = random.randint(0, n-1)
        visited = np.zeros(n)
        visited[curr] = 1
        tour = [curr]

        memory = {'s':[], 'a':[], 'logp':[], 'r':[], 'mask':[]}

        for _ in range(n-1):
            state = np.zeros(n*2)
            state[curr] = 1
            state[n:] = visited

            s_t = torch.FloatTensor(state).to(DEVICE)
            m_t = torch.FloatTensor(visited).to(DEVICE)

            with torch.no_grad():
                probs, _ = self.model(s_t, m_t)

            dist = Categorical(probs)
            action = dist.sample()

            memory['s'].append(state)
            memory['mask'].append(visited.copy())
            memory['a'].append(action.item())
            memory['logp'].append(dist.log_prob(action).item())

            next_node = action.item()
            r = -self.env.matrix[curr, next_node] # Immediate negative distance
            memory['r'].append(r)

            visited[next_node] = 1
            curr = next_node
            tour.append(curr)

        # Final loop closure cost
        total_dist = self.env.evaluate(tour)
        memory['r'][-1] -= self.env.matrix[tour[-1], tour[0]]

        return tour, total_dist, memory

    def _update_policy(self, memories):
        # Flatten batch
        states = torch.FloatTensor(np.array([m for traj in memories for m in traj['s']])).to(DEVICE)
        masks = torch.FloatTensor(np.array([m for traj in memories for m in traj['mask']])).to(DEVICE)
        actions = torch.LongTensor(np.array([m for traj in memories for m in traj['a']])).to(DEVICE)
        old_logps = torch.FloatTensor(np.array([m for traj in memories for m in traj['logp']])).to(DEVICE)

        # Returns
        all_returns = []
        for traj in memories:
            R = 0
            traj_ret = []
            for r in reversed(traj['r']):
                R = r + self.gamma * R
                traj_ret.insert(0, R)
            all_returns.extend(traj_ret)
        returns = torch.FloatTensor(all_returns).to(DEVICE)
        returns = (returns - returns.mean()) / (returns.std() + 1e-8)

        # PPO Step
        for _ in range(self.k_epochs):
            probs, values = self.model(states, masks)
            dist = Categorical(probs)
            logps = dist.log_prob(actions)
            entropy = dist.entropy().mean()

            ratio = torch.exp(logps - old_logps)
            surr1 = torch.clamp(ratio, 1-self.eps_clip, 1+self.eps_clip) * returns
            surr2 = torch.clamp(ratio, 1-self.eps_clip, 1+self.eps_clip) * returns
            loss = -torch.min(surr1, surr2).mean() - 0.01*entropy + 0.5*F.mse_loss(values.squeeze(), returns)

            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()

# ==========================================
# 2. LOCAL OPTIMIZATION: 2-OPT HEURISTIC
# ==========================================
def two_opt_local_search(route, matrix):
    """
    Applies 2-Opt local search to improve a single route.
    Swaps edges (i, i+1) and (j, j+1) if it reduces distance.
    """
    best_route = list(route) # Copy to avoid reference issues
    improved = True
    n = len(route)

    # Limit iterations for speed in a real-time loop
    max_iter = 100
    count = 0

    while improved and count < max_iter:
        improved = False
        count += 1
        for i in range(1, n - 2):
            for j in range(i + 1, n):
                if j - i == 1: continue

                # Check swap cost (Simplified delta evaluation)
                A, B = best_route[i-1], best_route[i]
                C, D = best_route[j], best_route[(j+1)%n]

                old_dist = matrix[A][B] + matrix[C][D]
                new_dist = matrix[A][C] + matrix[B][D]

                if new_dist < old_dist:
                    # Perform 2-opt swap (reverse segment i...j)
                    best_route[i:j+1] = best_route[i:j+1][::-1]
                    improved = True
    return best_route

# ==========================================
# 3. GENETIC ALGORITHM MODULE (Hybrid)
# ==========================================
class HybridGA:
    def __init__(self, env, pop_size=100, mutation_rate=0.05, use_2opt=True):
        self.env = env
        self.pop_size = pop_size
        self.rate = mutation_rate
        self.use_2opt = use_2opt

    def evolve(self, population, generations):
        history = []
        best_overall_route = None
        best_overall_val = float('inf')

        curr_pop = population
        print(f"  [GA Module] Evolving for {generations} generations (2-Opt={'On' if self.use_2opt else 'Off'})...")

        for g in range(generations):
            # 1. Evaluate
            scores = [self.env.evaluate(p) for p in curr_pop]
            min_s = min(scores)

            # Track Global Best
            if min_s < best_overall_val:
                best_overall_val = min_s
                best_overall_route = curr_pop[scores.index(min_s)]

                # FEEDBACK: Immediate Local Optimization on New Best
                if self.use_2opt:
                    improved = two_opt_local_search(best_overall_route, self.env.matrix)
                    new_val = self.env.evaluate(improved)
                    if new_val < best_overall_val:
                        best_overall_val = new_val
                        best_overall_route = improved

            history.append(best_overall_val)

            # 2. Selection (Roulette Wheel)
            # Avoid division by zero for perfect scores if any
            inv_scores = [1.0/(s + 1e-8) for s in scores]
            total = sum(inv_scores)
            probs = [i/total for i in inv_scores]

            # Elitism: Carry over top 2
            sorted_indices = np.argsort(scores)
            new_pop = [curr_pop[sorted_indices[0]], curr_pop[sorted_indices[1]]]

            # 3. Crossover & Mutation
            # Use `replace=True` as probabilities might lead to picking same parent multiple times
            parents_indices = np.random.choice(len(curr_pop), size=self.pop_size-2, p=probs, replace=True)
            parents = [curr_pop[i] for i in parents_indices]

            while len(new_pop) < self.pop_size:
                p1, p2 = random.sample(parents, 2)

                # Order Crossover (OX)
                n = self.env.num_cities
                start, end = sorted(random.sample(range(n), 2))
                child = [-1]*n
                child[start:end] = p1[start:end]
                ptr = end
                # Ensure that cities from p2 are added without duplicates and in order
                for city in p2:
                    if city not in child:
                        # Find the next available position for the city
                        while child[ptr] != -1:
                            ptr = (ptr + 1) % n
                        child[ptr] = city

                # Inversion Mutation
                if random.random() < self.rate:
                    i, j = sorted(random.sample(range(n), 2))
                    child[i:j] = reversed(child[i:j])

                # 4. Probabilistic 2-Opt (Local Optimization Feedback)
                # Apply 2-opt to 10% of children to maintain diversity while optimizing
                if self.use_2opt and random.random() < 0.1:
                    child = two_opt_local_search(child, self.env.matrix)

                new_pop.append(child)

            curr_pop = new_pop

        return best_overall_route, best_overall_val, history

# ==========================================
# 4. MAIN WORKFLOW & EVALUATION
# ==========================================
def print_matrix(res, optimal):
    print("\n" + "="*60)
    print("             IMPROVEMENT MATRIX & COMPARISON")
    print("="*60)

    df = pd.DataFrame({
        "Method": ["RL Module (Init)", "Standard GA (Baseline)", "Hybrid (RL+GA+2Opt)"],
        "Best Distance": [res["RL_Score"], res["Std_GA_Score"], res["Hybrid_Score"]],
        "Optimality Gap (%)": [f"{res['RL_Gap']:.2f}%", f"{res['Std_Gap']:.2f}%", f"{res['Hybrid_Gap']:.2f}%"]
    })
    print(df.to_string(index=False))
    print("-" * 60)
    print(f" Improvement over Pure RL:       {res['Imp_vs_RL']:.2f}%")
    print(f" Improvement over Standard GA:   {res['Imp_vs_Std']:.2f}%")
    print("="*60)

def plot_results(tsp, best_route, hist_hyb, hist_std, best_score, instance_name="TSP Instance"):
    plt.figure(figsize=(14, 6))

    # Plot 1: Convergence
    plt.subplot(1, 2, 1)
    plt.plot(hist_hyb, label='Hybrid (RL+GA+2Opt)', color='green', linewidth=2)
    plt.plot(hist_std, label='Standard GA', color='gray', linestyle='--')
    if tsp.optimal_value > 0: # Only plot optimal if it's a valid value
        plt.axhline(y=tsp.optimal_value, color='red', linestyle=':', label=f'Optimal ({tsp.optimal_value:.0f})')
    plt.xlabel("Generations")
    plt.ylabel("Tour Distance")
    plt.title(f"Convergence Comparison for {instance_name}")
    plt.legend()
    plt.grid(alpha=0.3)

    # Plot 2: Best Route Map
    plt.subplot(1, 2, 2)
    route = best_route + [best_route[0]] if best_route else [] # Handle empty route if problem occurs
    if route:
        coords = tsp.coords[route]
        plt.plot(coords[:,0], coords[:,1], 'o-', c='royalblue', markersize=5, label='Hybrid Route')
        plt.scatter(tsp.coords[0,0], tsp.coords[0,1], c='red', s=150, zorder=10, label='Start/End')

        # Annotate a few cities
        for i, txt in enumerate(range(len(tsp.coords))):
            if i % (len(tsp.coords) // 10 + 1) == 0: # Dynamically adjust annotation frequency
                plt.annotate(txt, (tsp.coords[i,0], tsp.coords[i,1]), xytext=(3, 3), textcoords='offset points')

    plt.title(f"Best Route for {instance_name} (Dist: {best_score:.2f})")
    plt.legend()
    plt.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

def plot_overall_comparisons(all_results):
    if not all_results:
        print("No results to plot for overall comparisons.")
        return

    # Prepare data for plotting
    instance_names = [res['filename'].replace('.tsp', '') for res in all_results]
    rl_gaps = [res['results']['RL_Gap'] for res in all_results]
    std_gaps = [res['results']['Std_Gap'] for res in all_results]
    hyb_gaps = [res['results']['Hybrid_Gap'] for res in all_results]

    imp_vs_rl = [res['results']['Imp_vs_RL'] for res in all_results]
    imp_vs_std = [res['results']['Imp_vs_Std'] for res in all_results]

    n_instances = len(instance_names)
    index = np.arange(n_instances)
    bar_width = 0.25

    # Plot 1: Optimality Gap Comparison
    plt.figure(figsize=(15, 7))
    plt.bar(index - bar_width, rl_gaps, bar_width, label='RL Gap (%)', color='skyblue')
    plt.bar(index, std_gaps, bar_width, label='Standard GA Gap (%)', color='lightcoral')
    plt.bar(index + bar_width, hyb_gaps, bar_width, label='Hybrid GA Gap (%)', color='mediumseagreen')

    plt.xlabel('TSP Instance')
    plt.ylabel('Optimality Gap (%)')
    plt.title('Optimality Gap Comparison Across TSP Instances')
    plt.xticks(index, instance_names, rotation=45, ha='right')
    plt.legend()
    plt.grid(axis='y', alpha=0.75)
    plt.tight_layout()
    plt.show()

    # Plot 2: Percentage Improvement of Hybrid GA
    plt.figure(figsize=(15, 7))
    plt.bar(index - bar_width/2, imp_vs_rl, bar_width, label='Hybrid GA Improvement over RL (%)', color='darkorange')
    plt.bar(index + bar_width/2, imp_vs_std, bar_width, label='Hybrid GA Improvement over Standard GA (%)', color='teal')

    plt.xlabel('TSP Instance')
    plt.ylabel('Percentage Improvement (%)')
    plt.title('Hybrid GA Performance Improvement Across TSP Instances')
    plt.xticks(index, instance_names, rotation=45, ha='right')
    plt.legend()
    plt.grid(axis='y', alpha=0.75)
    plt.tight_layout()
    plt.show()

def print_overall_summary(all_collected_results):
    if not all_collected_results:
        print("No results to summarize for overall comparison.")
        return

    # Collect data for summary
    rl_scores = []
    std_ga_scores = []
    hybrid_scores = []
    rl_gaps = []
    std_gaps = []
    hybrid_gaps = []
    imp_vs_rl = []
    imp_vs_std = []

    for res_data in all_collected_results:
        results = res_data['results']
        rl_scores.append(results['RL_Score'])
        std_ga_scores.append(results['Std_GA_Score'])
        hybrid_scores.append(results['Hybrid_Score'])
        rl_gaps.append(results['RL_Gap'])
        std_gaps.append(results['Std_Gap'])
        hybrid_gaps.append(results['Hybrid_Gap'])
        imp_vs_rl.append(results['Imp_vs_RL'])
        imp_vs_std.append(results['Imp_vs_Std'])

    # Calculate averages
    avg_rl_score = np.mean(rl_scores)
    avg_std_ga_score = np.mean(std_ga_scores)
    avg_hybrid_score = np.mean(hybrid_scores)

    avg_rl_gap = np.mean(rl_gaps)
    avg_std_ga_gap = np.mean(std_gaps)
    avg_hybrid_gap = np.mean(hybrid_gaps)

    avg_imp_vs_rl = np.mean(imp_vs_rl)
    avg_imp_vs_std = np.mean(imp_vs_std)

    print("\n" + "="*70)
    print("                 OVERALL SUMMARY ACROSS ALL INSTANCES")
    print("="*70)

    # Average Scores Table
    score_summary_df = pd.DataFrame({
        "Method": ["RL Module (Average)", "Standard GA (Average)", "Hybrid (Average)"],
        "Avg Best Distance": [f"{avg_rl_score:.2f}", f"{avg_std_ga_score:.2f}", f"{avg_hybrid_score:.2f}"]
    })
    print("\nAverage Best Distances:")
    print(score_summary_df.to_string(index=False))

    # Average Optimality Gaps Table
    gap_summary_df = pd.DataFrame({
        "Method": ["RL Module (Average)", "Standard GA (Average)", "Hybrid (Average)"],
        "Avg Optimality Gap (%)": [f"{avg_rl_gap:.2f}%", f"{avg_std_ga_gap:.2f}%", f"{avg_hybrid_gap:.2f}%"]
    })
    print("\nAverage Optimality Gaps:")
    print(gap_summary_df.to_string(index=False))

    # Average Improvement Table
    improvement_summary_df = pd.DataFrame({
        "Metric": ["Hybrid GA Improvement over RL (%)", "Hybrid GA Improvement over Standard GA (%)"],
        "Average Value": [f"{avg_imp_vs_rl:.2f}%", f"{avg_imp_vs_std:.2f}%"]
    })
    print("\nAverage Percentage Improvements of Hybrid GA:")
    print(improvement_summary_df.to_string(index=False))
    print("="*70)

def solve_tsp_instance(file_path=None):
    # Setup
    tsp = TSPManager(file_path)
    print(f"\nProblem Loaded: {tsp.num_cities} Cities. Optimal Target: {tsp.optimal_value}")

    # --- 1. RL MODULE (Generate Initial Pop) ---
    rl_agent = PPOAgent(tsp)
    # Generate high-quality seeds
    rl_population, rl_best_score = rl_agent.generate_population(population_size=100, episodes=400)
    print(f"  > RL Phase Complete. Best Score: {rl_best_score:.2f}")

    # --- 2. HYBRID GA (Refine Pop + 2-Opt) ---
    ga_hybrid = HybridGA(tsp, pop_size=100, use_2opt=True)
    # Initialize with RL seeds (padded with random if needed)
    pop_hybrid = deepcopy(rl_population)
    while len(pop_hybrid) < 100:
        p = list(range(tsp.num_cities))
        random.shuffle(p)
        pop_hybrid.append(p)

    hyb_tour, hyb_score, hyb_hist = ga_hybrid.evolve(pop_hybrid, generations=150)
    print(f"  > Hybrid GA Phase Complete. Best Score: {hyb_score:.2f}")

    # --- 3. BASELINE GA (Standard, Random Seed, No 2-Opt) ---
    print("\nRunning Baseline GA (Random Init, No 2-Opt) for Comparison...")
    ga_std = HybridGA(tsp, pop_size=100, use_2opt=False)
    pop_std = []
    for _ in range(100):
        p = list(range(tsp.num_cities))
        random.shuffle(p)
        pop_std.append(p)
    std_tour, std_score, std_hist = ga_std.evolve(pop_std, generations=150)

    # --- 4. IMPROVEMENT MATRIX & REPORT ---
    # Ensure optimal_value is not zero to avoid division by zero for gap calculation
    gap_rl = ((rl_best_score - tsp.optimal_value) / tsp.optimal_value) * 100 if tsp.optimal_value != 0 else float('inf')
    gap_std = ((std_score - tsp.optimal_value) / tsp.optimal_value) * 100 if tsp.optimal_value != 0 else float('inf')
    gap_hyb = ((hyb_score - tsp.optimal_value) / tsp.optimal_value) * 100 if tsp.optimal_value != 0 else float('inf')

    improve_vs_std = ((std_score - hyb_score) / std_score) * 100 if std_score != 0 else float('inf')
    improve_vs_rl = ((rl_best_score - hyb_score) / rl_best_score) * 100 if rl_best_score != 0 else float('inf')

    results = {
        "RL_Score": rl_best_score,
        "Std_GA_Score": std_score,
        "Hybrid_Score": hyb_score,
        "RL_Gap": gap_rl,
        "Std_Gap": gap_std,
        "Hybrid_Gap": gap_hyb,
        "Imp_vs_Std": improve_vs_std,
        "Imp_vs_RL": improve_vs_rl
    }

    print_matrix(results, tsp.optimal_value)

    # Return values for collection by run_full_pipeline
    return results, tsp, hyb_tour, hyb_hist, std_hist

def run_full_pipeline(directory_path='/content/'):
    # Initialize a list to store results from all instances
    all_collected_results = []
    print(f"\nScanning directory: {directory_path} for .tsp files...")

    # Iterate through all files in the specified directory
    for filename in os.listdir(directory_path):
        if filename.endswith('.tsp'):
            file_path = os.path.join(directory_path, filename)
            print(f"\n=== Processing TSP instance: {filename} ===")

            # Capture multiple returned values from solve_tsp_instance
            results, tsp_obj, hybrid_tour, hybrid_history, standard_history = solve_tsp_instance(file_path)

            # Collect results for overall comparison
            all_collected_results.append({
                "filename": filename,
                "results": results,
                "tsp_obj": tsp_obj,
                "hybrid_tour": hybrid_tour,
                "hybrid_history": hybrid_history,
                "standard_history": standard_history
            })

            # Plot results for the current instance
            plot_results(tsp_obj, hybrid_tour, hybrid_history, standard_history, results['Hybrid_Score'], instance_name=filename.replace('.tsp', ''))

    # Now, after processing all files, all_collected_results contains data for all instances.
    # You might want to return it or process it further here.
    return all_collected_results

if __name__ == "__main__":
    # When run as main, execute the pipeline and store the results
    overall_results = run_full_pipeline()
    # Call the new function to plot overall comparisons
    plot_overall_comparisons(overall_results)
    # Call the new function to print the overall summary matrix
    print_overall_summary(overall_results)
    print(f"\nOverall results collected and plotted for {len(overall_results)} instances.")
```

```python
# ================================
# COLAB NOTEBOOK: CLEAN PCB RUNS + STRONG VISUALS
# (Algorithm intact: PPO + GA + 2-Opt exactly as before)
# Adds: richer plots, summary charts, CSV export, and nicer reporting.
# ================================

!pip -q install tsplib95

import os, time, random
from copy import deepcopy

import tsplib95
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.distributions import Categorical

import matplotlib.pyplot as plt
import pandas as pd

# -------------------------
# 0) CONFIG
# -------------------------
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

PCB_DIR = "/content/pcb"
os.makedirs(PCB_DIR, exist_ok=True)

PCB_FILES = {
    "small":  "pcb442.tsp",
    "medium": "pcb1173.tsp",
    "large":  "pcb3038.tsp",
}

# TSPLIB published best-known/optimal tour lengths for these instances
# (Used only for GAP reporting; solver behavior unchanged)
KNOWN_OPTIMAL_VALUES = {
    "pcb442":  50778.0,
    "pcb1173": 56892.0,
    "pcb3038": 137694.0,
}

# Download source: TSPLIB mirror repo containing pcb442/pcb1173/pcb3038
BASE = "https://raw.githubusercontent.com/mastqe/tsplib/master"

# -------------------------
# 1) DOWNLOAD PCB DATASETS
# -------------------------
for size, fname in PCB_FILES.items():
    out = os.path.join(PCB_DIR, fname)
    if not os.path.exists(out) or os.path.getsize(out) == 0:
        !wget -q -O "{out}" "{BASE}/{fname}"
    print(f"{size:6} -> {out} (bytes={os.path.getsize(out)})")

# -------------------------
# 2) TSP ENV MANAGER
# -------------------------
class TSPManager:
    def __init__(self, filepath):
        if not (filepath and os.path.exists(filepath)):
            raise FileNotFoundError(f"Missing .tsp file: {filepath}")

        print(f"Loading: {filepath}")
        self.problem = tsplib95.load(filepath)

        self.num_cities = self.problem.dimension
        nodes = list(self.problem.get_nodes())
        self.coords = np.array([self.problem.node_coords[n] for n in nodes], dtype=np.float64)

        # Precompute Distance Matrix (unchanged)
        self.matrix = np.zeros((self.num_cities, self.num_cities), dtype=np.float64)
        for i in range(self.num_cities):
            for j in range(self.num_cities):
                if i != j:
                    self.matrix[i][j] = np.linalg.norm(self.coords[i] - self.coords[j])

        # Optimal for reporting
        name = str(self.problem.name).strip()
        self.optimal_value = KNOWN_OPTIMAL_VALUES.get(name, 0.0)
        if self.optimal_value == 0.0:
            print(f"Warning: No known TSPLIB best for '{name}'. GAP will be inf.")

    def evaluate(self, route):
        d = 0.0
        for i in range(len(route)-1):
            d += self.matrix[route[i], route[i+1]]
        d += self.matrix[route[-1], route[0]]
        return d

# -------------------------
# 3) RL MODULE: PPO (UNCHANGED)
# -------------------------
class PPOActorCritic(nn.Module):
    def __init__(self, num_cities, hidden_dim=128):
        super(PPOActorCritic, self).__init__()
        input_dim = num_cities * 2

        self.shared = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU()
        )
        self.actor = nn.Linear(hidden_dim, num_cities)
        self.critic = nn.Linear(hidden_dim, 1)

    def forward(self, x, mask):
        feat = self.shared(x)
        val = self.critic(feat)
        logits = self.actor(feat)
        logits = logits.masked_fill(mask.bool(), -1e9)
        probs = F.softmax(logits, dim=-1)
        return probs, val

class PPOAgent:
    def __init__(self, env, lr=1e-3):
        self.env = env
        self.model = PPOActorCritic(env.num_cities).to(DEVICE)
        self.optimizer = optim.AdamW(self.model.parameters(), lr=lr)
        self.gamma = 0.99
        self.eps_clip = 0.2
        self.k_epochs = 4

    def generate_population(self, population_size=100, episodes=400):
        print(f"  [RL Module] Training PPO for {episodes} episodes to generate population...")
        candidates = []
        best_rl_score = float('inf')

        for ep in range(episodes):
            tour, dist, memory = self._run_episode()

            if ep % 20 == 0 and ep > 0:
                self._update_policy([memory])

            candidates.append(tour)
            if dist < best_rl_score:
                best_rl_score = dist

            if ep % 100 == 0:
                print(f"    RL Episode {ep}: Best Score {best_rl_score:.2f}")

        candidates.sort(key=lambda x: self.env.evaluate(x))
        return candidates[:population_size], best_rl_score

    def _run_episode(self):
        n = self.env.num_cities
        curr = random.randint(0, n-1)
        visited = np.zeros(n)
        visited[curr] = 1
        tour = [curr]

        memory = {'s':[], 'a':[], 'logp':[], 'r':[], 'mask':[]}

        for _ in range(n-1):
            state = np.zeros(n*2)
            state[curr] = 1
            state[n:] = visited

            s_t = torch.FloatTensor(state).to(DEVICE)
            m_t = torch.FloatTensor(visited).to(DEVICE)

            with torch.no_grad():
                probs, _ = self.model(s_t, m_t)

            dist = Categorical(probs)
            action = dist.sample()

            memory['s'].append(state)
            memory['mask'].append(visited.copy())
            memory['a'].append(action.item())
            memory['logp'].append(dist.log_prob(action).item())

            next_node = action.item()
            r = -self.env.matrix[curr, next_node]
            memory['r'].append(r)

            visited[next_node] = 1
            curr = next_node
            tour.append(curr)

        total_dist = self.env.evaluate(tour)
        memory['r'][-1] -= self.env.matrix[tour[-1], tour[0]]
        return tour, total_dist, memory

    def _update_policy(self, memories):
        states = torch.FloatTensor(np.array([m for traj in memories for m in traj['s']])).to(DEVICE)
        masks  = torch.FloatTensor(np.array([m for traj in memories for m in traj['mask']])).to(DEVICE)
        actions = torch.LongTensor(np.array([m for traj in memories for m in traj['a']])).to(DEVICE)
        old_logps = torch.FloatTensor(np.array([m for traj in memories for m in traj['logp']])).to(DEVICE)

        all_returns = []
        for traj in memories:
            R = 0
            traj_ret = []
            for r in reversed(traj['r']):
                R = r + self.gamma * R
                traj_ret.insert(0, R)
            all_returns.extend(traj_ret)
        returns = torch.FloatTensor(all_returns).to(DEVICE)
        returns = (returns - returns.mean()) / (returns.std() + 1e-8)

        for _ in range(self.k_epochs):
            probs, values = self.model(states, masks)
            dist = Categorical(probs)
            logps = dist.log_prob(actions)
            entropy = dist.entropy().mean()

            ratio = torch.exp(logps - old_logps)
            surr1 = torch.clamp(ratio, 1-self.eps_clip, 1+self.eps_clip) * returns
            surr2 = torch.clamp(ratio, 1-self.eps_clip, 1+self.eps_clip) * returns
            loss = -torch.min(surr1, surr2).mean() - 0.01*entropy + 0.5*F.mse_loss(values.squeeze(), returns)

            self.optimizer.zero_grad()
            loss.backward()
            self.optimizer.step()

# -------------------------
# 4) 2-OPT (UNCHANGED)
# -------------------------
def two_opt_local_search(route, matrix):
    best_route = list(route)
    improved = True
    n = len(route)
    max_iter = 100
    count = 0

    while improved and count < max_iter:
        improved = False
        count += 1
        for i in range(1, n - 2):
            for j in range(i + 1, n):
                if j - i == 1:
                    continue
                A, B = best_route[i-1], best_route[i]
                C, D = best_route[j], best_route[(j+1)%n]
                old_dist = matrix[A][B] + matrix[C][D]
                new_dist = matrix[A][C] + matrix[B][D]
                if new_dist < old_dist:
                    best_route[i:j+1] = best_route[i:j+1][::-1]
                    improved = True
    return best_route

# -------------------------
# 5) GA (UNCHANGED)
# -------------------------
class HybridGA:
    def __init__(self, env, pop_size=100, mutation_rate=0.05, use_2opt=True):
        self.env = env
        self.pop_size = pop_size
        self.rate = mutation_rate
        self.use_2opt = use_2opt

    def evolve(self, population, generations):
        history = []
        best_overall_route = None
        best_overall_val = float('inf')

        curr_pop = population
        print(f"  [GA Module] Evolving for {generations} generations (2-Opt={'On' if self.use_2opt else 'Off'})...")

        for g in range(generations):
            scores = [self.env.evaluate(p) for p in curr_pop]
            min_s = min(scores)

            if min_s < best_overall_val:
                best_overall_val = min_s
                best_overall_route = curr_pop[scores.index(min_s)]

                if self.use_2opt:
                    improved = two_opt_local_search(best_overall_route, self.env.matrix)
                    new_val = self.env.evaluate(improved)
                    if new_val < best_overall_val:
                        best_overall_val = new_val
                        best_overall_route = improved

            history.append(best_overall_val)

            inv_scores = [1.0/(s + 1e-8) for s in scores]
            total = sum(inv_scores)
            probs = [i/total for i in inv_scores]

            sorted_indices = np.argsort(scores)
            new_pop = [curr_pop[sorted_indices[0]], curr_pop[sorted_indices[1]]]

            parents_indices = np.random.choice(len(curr_pop), size=self.pop_size-2, p=probs, replace=True)
            parents = [curr_pop[i] for i in parents_indices]

            while len(new_pop) < self.pop_size:
                p1, p2 = random.sample(parents, 2)

                n = self.env.num_cities
                start, end = sorted(random.sample(range(n), 2))
                child = [-1]*n
                child[start:end] = p1[start:end]
                ptr = end
                for city in p2:
                    if city not in child:
                        while child[ptr] != -1:
                            ptr = (ptr + 1) % n
                        child[ptr] = city

                if random.random() < self.rate:
                    i, j = sorted(random.sample(range(n), 2))
                    child[i:j] = reversed(child[i:j])

                if self.use_2opt and random.random() < 0.1:
                    child = two_opt_local_search(child, self.env.matrix)

                new_pop.append(child)

            curr_pop = new_pop

        return best_overall_route, best_overall_val, history

# -------------------------
# 6) REPORTING + VISUALS
# -------------------------
def print_matrix(res, optimal):
    print("\n" + "="*78)
    print(f"IMPROVEMENT MATRIX | {res['Instance']} | N={res['N']} | Runtime={res['Runtime_s']:.2f}s")
    print("="*78)
    df = pd.DataFrame({
        "Method": ["RL Module (Init)", "Standard GA (Baseline)", "Hybrid (RL+GA+2Opt)"],
        "Best Distance": [res["RL_Score"], res["Std_GA_Score"], res["Hybrid_Score"]],
        "Optimality Gap (%)": [f"{res['RL_Gap']:.2f}%", f"{res['Std_Gap']:.2f}%", f"{res['Hybrid_Gap']:.2f}%"]
    })
    print(df.to_string(index=False))
    print("-"*78)
    print(f" Improvement over Pure RL:       {res['Imp_vs_RL']:.2f}%")
    print(f" Improvement over Standard GA:   {res['Imp_vs_Std']:.2f}%")
    print("="*78)

def plot_convergence(hyb_hist, std_hist, optimal_value, title):
    plt.figure(figsize=(10, 5))
    plt.plot(hyb_hist, label="Hybrid (RL+GA+2Opt)", linewidth=2)
    plt.plot(std_hist, label="Standard GA", linestyle="--")
    if optimal_value > 0:
        plt.axhline(y=optimal_value, linestyle=":", label=f"TSPLIB best ({optimal_value:.0f})")
    plt.xlabel("Generations")
    plt.ylabel("Tour Distance")
    plt.title(title)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()

def plot_route_2d(tsp, route, title):
    # NOTE: for large instances, plotting every point can be heavy.
    # We still plot it (no algorithm impact). If it feels slow, skip route plot for pcb3038.
    if route is None or len(route) == 0:
        print("No route to plot.")
        return

    plt.figure(figsize=(7, 7))
    r = route + [route[0]]
    coords = tsp.coords[r]
    plt.plot(coords[:,0], coords[:,1], 'o-', markersize=2)
    plt.scatter(tsp.coords[r[0],0], tsp.coords[r[0],1], s=80, zorder=10)
    plt.title(title)
    plt.grid(alpha=0.3)
    plt.show()

def plot_overall_comparisons(df):
    # df columns expected:
    # Size, Instance, N, Runtime_s, RL_Score, Std_GA_Score, Hybrid_Score,
    # RL_Gap, Std_Gap, Hybrid_Gap, Imp_vs_RL, Imp_vs_Std

    df2 = df.copy()
    df2["Runtime_min"] = df2["Runtime_s"] / 60.0

    labels = df2["Instance"].tolist()
    x = np.arange(len(labels))
    w = 0.25

    # ---- Plot A: Best distance (3 methods)
    plt.figure(figsize=(13, 6))
    plt.bar(x - w, df2["RL_Score"], width=w, label="RL Init")
    plt.bar(x,      df2["Std_GA_Score"], width=w, label="Standard GA")
    plt.bar(x + w,  df2["Hybrid_Score"], width=w, label="Hybrid (RL+GA+2Opt)")
    plt.xticks(x, labels, rotation=0)
    plt.ylabel("Best Distance")
    plt.title("Best Distance Comparison (PCB Instances)")
    plt.grid(axis="y", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # ---- Plot B: Optimality Gap (%)
    plt.figure(figsize=(13, 6))
    plt.bar(x - w, df2["RL_Gap"], width=w, label="RL Gap (%)")
    plt.bar(x,     df2["Std_Gap"], width=w, label="Standard GA Gap (%)")
    plt.bar(x + w, df2["Hybrid_Gap"], width=w, label="Hybrid Gap (%)")
    plt.xticks(x, labels, rotation=0)
    plt.ylabel("Optimality Gap (%)")
    plt.title("Optimality Gap Comparison (Lower is Better)")
    plt.grid(axis="y", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # ---- Plot C: Improvement (%)
    plt.figure(figsize=(13, 6))
    plt.bar(x - w/2, df2["Imp_vs_RL"], width=w, label="Hybrid Improvement vs RL (%)")
    plt.bar(x + w/2, df2["Imp_vs_Std"], width=w, label="Hybrid Improvement vs Standard GA (%)")
    plt.xticks(x, labels, rotation=0)
    plt.ylabel("Improvement (%)")
    plt.title("Hybrid Improvements Across PCB Sizes")
    plt.grid(axis="y", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # ---- Plot D: Runtime (minutes)
    plt.figure(figsize=(13, 5))
    plt.bar(x, df2["Runtime_min"], width=0.5)
    plt.xticks(x, labels, rotation=0)
    plt.ylabel("Runtime (minutes)")
    plt.title("Runtime per Instance (End-to-End)")
    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

# -------------------------
# 7) SOLVER WRAPPER (UNCHANGED COMPUTE; adds logging + returns more artifacts)
# -------------------------
def solve_one(file_path, episodes=400, generations=150, pop_size=100, plot_route=True):
    tsp = TSPManager(file_path)
    print(f"Problem: {tsp.problem.name} | N={tsp.num_cities} | TSPLIB best={tsp.optimal_value}")

    t0 = time.time()

    # RL seeds
    rl_agent = PPOAgent(tsp)
    rl_population, rl_best_score = rl_agent.generate_population(population_size=pop_size, episodes=episodes)
    print(f"  > RL done. Best score: {rl_best_score:.2f}")

    # Hybrid GA
    ga_hybrid = HybridGA(tsp, pop_size=pop_size, use_2opt=True)
    pop_hybrid = deepcopy(rl_population)
    while len(pop_hybrid) < pop_size:
        p = list(range(tsp.num_cities))
        random.shuffle(p)
        pop_hybrid.append(p)

    hyb_tour, hyb_score, hyb_hist = ga_hybrid.evolve(pop_hybrid, generations=generations)
    print(f"  > Hybrid GA done. Best score: {hyb_score:.2f}")

    # Baseline GA
    print("  Running baseline GA (random init, no 2-opt)...")
    ga_std = HybridGA(tsp, pop_size=pop_size, use_2opt=False)
    pop_std = []
    for _ in range(pop_size):
        p = list(range(tsp.num_cities))
        random.shuffle(p)
        pop_std.append(p)
    std_tour, std_score, std_hist = ga_std.evolve(pop_std, generations=generations)

    runtime = time.time() - t0

    # Gaps
    if tsp.optimal_value != 0:
        gap_rl  = ((rl_best_score - tsp.optimal_value) / tsp.optimal_value) * 100
        gap_std = ((std_score    - tsp.optimal_value) / tsp.optimal_value) * 100
        gap_hyb = ((hyb_score    - tsp.optimal_value) / tsp.optimal_value) * 100
    else:
        gap_rl = gap_std = gap_hyb = float("inf")

    improve_vs_std = ((std_score - hyb_score) / std_score) * 100 if std_score != 0 else float("inf")
    improve_vs_rl  = ((rl_best_score - hyb_score) / rl_best_score) * 100 if rl_best_score != 0 else float("inf")

    results = {
        "Instance": str(tsp.problem.name),
        "N": tsp.num_cities,
        "Runtime_s": runtime,
        "RL_Score": rl_best_score,
        "Std_GA_Score": std_score,
        "Hybrid_Score": hyb_score,
        "RL_Gap": gap_rl,
        "Std_Gap": gap_std,
        "Hybrid_Gap": gap_hyb,
        "Imp_vs_Std": improve_vs_std,
        "Imp_vs_RL": improve_vs_rl,
        "TSPLIB_Best": tsp.optimal_value
    }

    print_matrix(results, tsp.optimal_value)

    # Per-instance visuals
    plot_convergence(
        hyb_hist, std_hist, tsp.optimal_value,
        title=f"Convergence: {tsp.problem.name} (N={tsp.num_cities})"
    )

    if plot_route:
        # Route plotting can be heavy for pcb3038; can disable by passing plot_route=False
        plot_route_2d(
            tsp, hyb_tour,
            title=f"Best Hybrid Route: {tsp.problem.name} | Dist={hyb_score:.2f}"
        )

    return results, hyb_hist, std_hist

# -------------------------
# 8) RUN PCB: SMALL / MEDIUM / LARGE
# -------------------------
all_results = []
per_instance_histories = {}  # store convergence curves

for size, fname in PCB_FILES.items():
    path = os.path.join(PCB_DIR, fname)
    print("\n" + "#"*88)
    print(f"RUN: {size.upper()}  ->  {fname}")
    print("#"*88)

    # For the largest instance, route plotting can be slow. Keep convergence plots on.
    plot_route_flag = (size != "large")

    res, hyb_hist, std_hist = solve_one(path, episodes=400, generations=150, pop_size=100, plot_route=plot_route_flag)
    res["Size"] = size
    all_results.append(res)

    per_instance_histories[res["Instance"]] = {
        "hyb_hist": hyb_hist,
        "std_hist": std_hist,
        "tsplib_best": res["TSPLIB_Best"]
    }

df = pd.DataFrame(all_results)
df = df[[
    "Size","Instance","N","Runtime_s","TSPLIB_Best",
    "RL_Score","Std_GA_Score","Hybrid_Score",
    "RL_Gap","Std_Gap","Hybrid_Gap",
    "Imp_vs_RL","Imp_vs_Std"
]]

print("\n\nFINAL PCB SUMMARY:")
print(df.to_string(index=False))

# Save CSV for report
out_csv = "/content/pcb_results_summary.csv"
df.to_csv(out_csv, index=False)
print("\nSaved:", out_csv)

# -------------------------
# 9) OVERALL COMPARISON GRAPHS (MANY)
# -------------------------
plot_overall_comparisons(df)

# Extra: overlay convergence curves for Hybrid only (all instances together)
plt.figure(figsize=(12, 5))
for inst, h in per_instance_histories.items():
    plt.plot(h["hyb_hist"], label=f"{inst} Hybrid")
plt.xlabel("Generations")
plt.ylabel("Tour Distance")
plt.title("Hybrid Convergence Overlay (All PCB Instances)")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Extra: overlay convergence curves for Standard GA only (all instances together)
plt.figure(figsize=(12, 5))
for inst, h in per_instance_histories.items():
    plt.plot(h["std_hist"], label=f"{inst} Standard GA", linestyle="--")
plt.xlabel("Generations")
plt.ylabel("Tour Distance")
plt.title("Standard GA Convergence Overlay (All PCB Instances)")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Extra: Scatter of N vs Hybrid Gap
plt.figure(figsize=(8, 5))
plt.scatter(df["N"], df["Hybrid_Gap"])
for i in range(len(df)):
    plt.annotate(df.loc[i, "Instance"], (df.loc[i, "N"], df.loc[i, "Hybrid_Gap"]), xytext=(5,5), textcoords="offset points")
plt.xlabel("Number of Nodes (N)")
plt.ylabel("Hybrid Optimality Gap (%)")
plt.title("Scaling View: N vs Hybrid Gap")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nAll plots generated. CSV saved for report.")
```
